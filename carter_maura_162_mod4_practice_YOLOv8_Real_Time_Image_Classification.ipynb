**YOLOv8 Classification Overview**

YOLOv8 extends the YOLO family to image classification tasks. Key features include:


**1. Lightweight “nano” variant** (yolov8n-cls.pt) for fast inference

**2. PyTorch-based** implementation with a simple Python API

3. Outputs a probability distribution over 1,000 ImageNet-style classes

*This model is ideal for real-time on-device scenarios or as a baseline for transfer learning.*

**Google Colab Demo: Classifying a Cat Image**

Copy the code below into a Colab notebook (Runtime ▶ Change runtime type ▶ GPU), then run each cell.

**1. Install Ultralytics YOLOv8**

```python
#python
# 1. Install Ultralytics YOLOv8
!pip install -q ultralytics
```

**2. Imports**

```python
# 2. Imports
from ultralytics import YOLO      # model API
import torch                      # tensor operations
import numpy as np               # array handling
```

**3. Load the YOLOv8 classification model**

```python
# 3. Load the YOLOv8 classification model
#    - 'yolov8n-cls.pt' is the nano variant pretrained on ImageNet-like data
model_cls = YOLO('yolov8n-cls.pt')
```


**4. Run inference on a publicly-hosted cat image**

```python
# 4. Run inference on a publicly-hosted cat image
#    - Passing the URL directly returns a list of Results objects
results = model_cls.predict(
    source='https://upload.wikimedia.org/wikipedia/commons/3/3a/Cat03.jpg'
)
```

**5. Extract the first (and only) Results object**

```python
# 5. Extract the first (and only) Results object
res = results[0]
```

**6. Access the raw probability tensor**

```python
# 6. Access the raw probability tensor
#    - res.probs is a Probs wrapper; .data holds the actual torch.Tensor
raw_tensor = res.probs.data
```

**7. Convert to a flat NumPy array**

```python
# 7. Convert to a flat NumPy array
probs_np = raw_tensor.detach().cpu().numpy().flatten()
```

**8. Determine Top-1 prediction**

```python
# 8. Determine Top-1 prediction
top1_idx  = int(np.argmax(probs_np))
top1_name = model_cls.names[top1_idx]
top1_conf = probs_np[top1_idx]
```

**9. Determine Top-5 predictions**

```python
# 9. Determine Top-5 predictions
top5_idxs = np.argsort(probs_np)[-5:][::-1]
top5      = [(model_cls.names[i], probs_np[i]) for i in top5_idxs]
```

**10. Display the results**

```python
# 10. Display the results
print(f"\nTop-1 Prediction: {top1_name} ({top1_conf*100:.2f}%)\n")
print("Top-5 Predictions:")
for rank, (name, conf) in enumerate(top5, start=1):
    print(f" {rank}. {name} ({conf*100:.2f}%)")
```

**Component Breakdown**

Printing loop Iterates through the Top-5 list, formatting rank, class name, and confidence percentage.
